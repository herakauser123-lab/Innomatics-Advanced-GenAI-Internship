##Problem Statement 1:Smart Parking Lot Management System

```python
# Function to manage smart parking system
def manage_parking(capacity, logs):

    current_vehicles = 0
    peak_vehicles = 0

    # Process vehicle logs
    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            current_vehicles -= 1

        # Track peak usage
        if current_vehicles > peak_vehicles:
            peak_vehicles = current_vehicles

    # Check parking status
    if current_vehicles > capacity:
        status = "Over Capacity"
    else:
        status = "Available"

    print("Currently Parked Vehicles:", current_vehicles)
    print("Peak Parking Usage:", peak_vehicles)
    print("Parking Status:", status)


# Sample Input
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

manage_parking(capacity, vehicle_logs)
```

##Problem Statement 2:Online Food Delivery Time Estimator

```python
def delivery_time(distance, traffic, whether):
  #base time is 5 minutes per km
  time = distance * 5

  #traffic delay
  if traffic == "high":
    time += 10
  elif traffic == "medium":
    time += 5

  #whether delay
  if whether == "rainy":
    time += 5

  print("Estimated Delivery Time : ",time," Minutes")


delivery_time(8, "high", "rainy")
```

##Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

```python
# Function to analyze seat occupancy
def analyze_occupancy(total_seats, booked_seats):

    booked_count = len(booked_seats)

    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100

    # Determine show status
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    # Suggest additional show
    if occupancy >= 90:
        suggestion = "Consider Opening Additional Show"
    else:
        suggestion = "No Additional Show Needed"

    print("Occupancy:", occupancy, "%")
    print("Show Status:", status)
    print("Suggestion:", suggestion)


# Sample Input
total_seats = 200
booked_seats = [1] * 150   # 150 booked seats

analyze_occupancy(total_seats, booked_seats)
```

##Problem Statement 4:Cloud Server Load Classification System

```python
# Function to classify server load
def classify_server_load(cpu_readings):


    # Calculate average
    average = sum(cpu_readings) / len(cpu_readings)

    # Determine server status
    if average < 50:
        status = "Normal"
    elif average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Average CPU Load:", average, "%")
    print("Server Status:", status)


# Sample Input
readings = [45, 60, 70, 85, 90]
classify_server_load(readings)
```

## Problem Statement 5:Smart Classroom Resource Usage Monitor

```python
# Function to monitor classroom resource usage
def monitor_resources(resource_usage):

    overused = []

    # Assume overuse threshold is more than 8 hours
    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)

    # Print results
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resources(usage)
```

## Problem Statement 6: Online Event Registration Capacity Controller

```python
# Function to control event registration capacity
def manage_registrations(capacity, registrations):

    confirmed = 0
    waitlisted = 0

    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1

    status = "Closed" if confirmed == capacity else "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Input
manage_registrations(100, 105)
```
