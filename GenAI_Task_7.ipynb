##Problem 1: Smart Light Controller

```python
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"

    def turn_on(self):
        self.status = "ON"
        print(f"{self.name} is {self.status}")

    def turn_off(self):
        self.status = "OFF"
        print(f"{self.name} is {self.status}")

    def display_status(self):
        print(f"{self.name} is {self.status}")


# Sample Input
light_name = "Bedroom Light"
action = "ON"

# Create SmartLight object
light = SmartLight(light_name)

# Perform action
if action.upper() == "ON":
    light.turn_on()
elif action.upper() == "OFF":
    light.turn_off()
else:
    print("Invalid Action")
```

##Problem 2: Employee ID Card System

```python
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")


# Sample Input
name = "Rahul"
emp_id = "EMP102"
department = "IT"

# Create Employee object
employee = Employee(name, emp_id, department)

# Display ID Card
employee.display_id_card()
```

##Problem 3: Mobile Contact Record

```python
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def save_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# Sample Input
name = "Anita"
phone = "9876543210"

# Create Contact object
contact = Contact(name, phone)

# Display Contact Information
contact.save_contact()
```

##Problem 4: Product Price Tag Generator

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def print_price_tag(self):
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")


# Sample Input
product_name = "Headphones"
price = 2499

# Create Product object
product = Product(product_name, price)

# Display Price Tag
product.print_price_tag()
```

##Problem 5: Movie Rating Display System

```python
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_details(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")


# Sample Input
movie_name = "Inception"
rating = 4.8

# Create Movie object
movie = Movie(movie_name, rating)

# Display Movie Details
movie.display_details()
```

##Problem 6: Delivery Address Manager

```python
class Delivery:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def print_delivery_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")


# Sample Input
customer_name = "Suman"
address = "Hyderabad"

# Create Delivery object
delivery = Delivery(customer_name, address)

# Display Delivery Details
delivery.print_delivery_details()
```
